```python
plt.figure(figsize=(6,4))
plt.bar(df["type"].value_counts().index,
       df["type"].value_counts().values)
plt.xlabel("content type")
plt.ylabel("count")
plt.title("movies VS TV shows")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(df["release_year"].value_counts().index,
       df["release_year"].value_counts().values)
plt.xlabel("release year")
plt.ylabel("number of content")
plt.title("release_year trend")
plt.show()
```

```python
country=df["country"].value_counts().head(10)
plt.figure(figsize=(6,4))
plt.bar(country.index,
       country.values)
plt.xlabel("country")
plt.ylabel("number of content")
plt.title("country VS content")
plt.xticks(rotation=45)
plt.show()

```

```python
Rating=df["rating"].value_counts().head(10)
plt.figure(figsize=(6,4))
plt.bar(Rating.index,
       Rating.values)
plt.xlabel("Rating")
plt.ylabel("number of content")
plt.title("Rating wise content distribution")
plt.xticks(rotation=45)
plt.show()

```

```python
plt.figure(figsize=(6,4))
plt.hist(df["duration_minutes"],bins=20)
plt.xlabel("duration(minutes)")
plt.ylabel("number of mivies")
plt.title("duration VS movies")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["duration_sessions"],bins=20)
plt.xlabel("duration(sessions)")
plt.ylabel("number of sessions")
plt.title("number of sessions")
plt.show()
```

```python


content= df["type"].value_counts()

plt.figure(figsize=(6,4))
plt.pie(content.values,
    labels=content.index,
startangle=90,
autopct="%1.1f%%")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["release_year"],df["duration_sessions"])
plt.xlabel("release_year")
plt.ylabel("duration_sessions")
plt.title("Relationship Between Release Year and Duration")
plt.show()
```
